### 1.1. Probabilistic Queries: Shapes and Counts

Consider a full joint distribution over four variables $(A, B, C, D)$ with cardinalities $|A|=2$, $|B|=3$, $|C|=5$, and $|D|=5$.

For each query below:

1. **Provide the shape** of the answer to the query.
   - Express the shape as a tuple of dimension sizes, in the same order as the variables in the query. Omit singleton dimensions.
   - Example: if the query is $P(D, A=1)$, the expected shape is `(5,)`.
2. **Provide the total number of non-redundant values** in the result.
3. **Provide the number of probability distributions** in the result (the number of complete conditional distributions that are “stacked together”). 

Queries:
1. $P(A \mid B)$  
2. $P(C, D \mid A=1, B=0)$  
3. $P(B \mid C=4)$  
4. $P(A=1 \mid D=4)$

```python
result = {
    "P(A|B)": {"shape": (2,3), "n_values": 6, "n_distributions": 3},
    "P(C,D|A=1,B=0)": {"shape": (5, 5), "n_values": 25, "n_distributions": 1},
    "P(B|C=4)": {"shape": (3,), "n_values": 3, "n_distributions": 1},
    "P(A=1|D=4)": {"shape": (), "n_values": 1, "n_distributions": 1},
}
```

```python
def pt(k, f):
    if k not in result: raise AssertionError(f"Missing {k}")
    if f not in result[k]: raise AssertionError(f"{k}:{f} missing")
    v = result[k][f]
    if v is None: raise AssertionError(f"{k}:{f} is None")
```

```python
# format check
for f in ("shape","n_values","n_distributions"): pt("P(A|B)", f)
```

```python
for f in ("shape","n_values","n_distributions"): pt("P(C,D|A=1,B=0)", f)
```

```python
for f in ("shape","n_values","n_distributions"): pt("P(B|C=4)", f)
```

```python
for f in ("shape","n_values","n_distributions"): pt("P(A=1|D=4)", f)
```

### 1.2. Counting Arithmetic Operations: Inference via Enumeration vs. Variable Elimination

We continue with the same FJDT over $(A,B,C,D)$ (cardinalities $|A|=2$, $|B|=3$, $|C|=5$, $|D|=5$) which factorizes into the following conditional probability distributions:
- $P(A)$
- $P(B \mid A)$
- $P(C \mid A)$
- $P(D \mid B C)$

For each query below, report the **minimum** number of multiplication operations needed for inference via **Enumeration** and **Variable Elimination**.
- `n_op_enum`: the total number of **additions and multiplications** for a naive implementaiton of inference by enumeration which constructs atomic probabilites from the conditional probability distribution tables and sum over the hidden variables.
- `n_op_elim`: the total number of **additions and multiplications** for inference by variable elimination.

Queries:
1. $P(A \mid B=\lnot b)$  
2. $P(A=1 \mid D=4,\; C=0)$  
3. $P(A,\; D \mid B,\; C)$

```python
op_min = {
    "P(A|B=not b)":      {"n_op_enum": 225, "n_op_elim": 125},
    "P(A=1|D=4,C=0)":    {"n_op_enum": 24,  "n_op_elim": 16},
    "P(A,D|B,C)":        {"n_op_enum": 471,  "n_op_elim": 471},
}
```

```python
# shared helpers
def pt(k, f):
    if k not in op_min: raise AssertionError(f"Missing {k}")
    if f not in op_min[k]: raise AssertionError(f"{k}:{f} missing")
    v = op_min[k][f]
    if v is None: raise AssertionError(f"{k}:{f} is None")
    if type(v) is not int: raise AssertionError(f"{k}:{f} must be int")
```

```python
pt("P(A|B=not b)", "n_op_enum")
```

```python
pt("P(A|B=not b)", "n_op_elim")
```

```python
pt("P(A=1|D=4,C=0)", "n_op_enum")
```

```python
pt("P(A=1|D=4,C=0)", "n_op_elim")
```

```python
pt("P(A,D|B,C)", "n_op_enum")
```

```python
pt("P(A,D|B,C)", "n_op_elim")
```
